```python
# =========================================================
# MULTI-CHAIN TOKEN ANALYZER
# READ-ONLY — NO TRADING
# =========================================================

def analyze_token(chain, token):

    if chain not in CHAINS:
        raise Exception(
            "Unsupported chain. Use: solana, bnb, or robinhood."
        )

    pairs = get_pairs(chain, token)

    valid_pairs = []

    for p in pairs:

        liquidity = (
            p.get("liquidity", {})
             .get("usd")
        )

        if liquidity is not None:
            valid_pairs.append(p)

    if not valid_pairs:
        raise Exception(
            "No usable liquidity/pair data found."
        )

    # Sort by liquidity
    valid_pairs.sort(
        key=lambda p:
        p.get("liquidity", {}).get("usd", 0) or 0,
        reverse=True
    )

    total_liquidity = sum(
        p.get("liquidity", {}).get("usd", 0) or 0
        for p in valid_pairs
    )

    total_volume = sum(
        p.get("volume", {}).get("h24", 0) or 0
        for p in valid_pairs
    )

    buys = 0
    sells = 0

    for p in valid_pairs:

        txns = p.get("txns", {}).get("h24", {})

        buys += txns.get("buys", 0) or 0
        sells += txns.get("sells", 0) or 0

    trades = buys + sells

    buy_pressure = (
        buys / trades * 100
        if trades
        else 0
    )

    volume_liquidity = (
        total_volume / total_liquidity
        if total_liquidity
        else 0
    )

    strongest = valid_pairs[0]

    name = strongest.get(
        "baseToken", {}
    ).get("name", "Unknown")

    symbol = strongest.get(
        "baseToken", {}
    ).get("symbol", "Unknown")

    price = strongest.get(
        "priceUsd",
        "Unknown"
    )

    # -----------------------------
    # BASIC RISK SCORE
    # -----------------------------

    score = 100
    flags = []

    if total_liquidity < 5000:

        score -= 35
        flags.append(
            "🔴 Very low liquidity"
        )

    elif total_liquidity < 10000:

        score -= 20
        flags.append(
            "🟠 Low liquidity"
        )

    elif total_liquidity < 50000:

        score -= 10
        flags.append(
            "🟡 Moderate liquidity"
        )

    if volume_liquidity > 30:

        score -= 20
        flags.append(
            "🟠 Extremely high volume/liquidity"
        )

    elif volume_liquidity > 10:

        score -= 10
        flags.append(
            "🟡 High volume/liquidity"
        )

    if trades:

        if buy_pressure < 45:

            score -= 20
            flags.append(
                "🔴 Strong sell pressure"
            )

        elif buy_pressure < 50:

            score -= 10
            flags.append(
                "🟠 Sell pressure"
            )

    score = max(0, min(100, score))

    if score >= 80:

        rating = "🟢 LOWER RISK — INVESTIGATE"

    elif score >= 60:

        rating = "🟡 MODERATE RISK"

    elif score >= 40:

        rating = "🟠 HIGH RISK"

    else:

        rating = "🔴 EXTREME RISK"

    return {
        "chain": chain,
        "token": token,
        "name": name,
        "symbol": symbol,
        "price": price,
        "liquidity": total_liquidity,
        "volume": total_volume,
        "buys": buys,
        "sells": sells,
        "buy_pressure": buy_pressure,
        "volume_liquidity": volume_liquidity,
        "pools": len(valid_pairs),
        "score": score,
        "rating": rating,
        "flags": flags,
        "dex": strongest.get("dexId", "Unknown"),
        "pair": strongest.get(
            "pairAddress",
            "Unknown"
        ),
        "url": strongest.get("url", "")
    }


print("✅ Multi-chain analyzer loaded.")
print("🔐 READ-ONLY")
```

```python
# =========================================================
# TEMMYDEEXC — PASTE-ONLY TELEGRAM SCANNER
# READ-ONLY — NO BUYING / SELLING
# =========================================================

import requests
import time
import re

def send_message(chat_id, text):
    requests.post(
        f"https://api.telegram.org/bot{TELEGRAM_TOKEN}/sendMessage",
        json={
            "chat_id": chat_id,
            "text": text,
            "disable_web_page_preview": True
        },
        timeout=20
    )


def looks_like_solana(address):
    # Solana addresses are normally 32–44 characters,
    # using Base58 characters.
    return bool(
        re.fullmatch(r"[1-9A-HJ-NP-Za-km-z]{32,44}", address)
    )


def looks_like_evm(address):
    return bool(
        re.fullmatch(r"0x[a-fA-F0-9]{40}", address)
    )


def scan_automatic(token):

    # -----------------------------------------------------
    # SOLANA
    # -----------------------------------------------------

    if looks_like_solana(token):

        print(f"🔎 Checking SOLANA: {token}")

        try:
            result = analyze_token(
                "solana",
                token
            )

            return make_alert(result)

        except Exception as e:

            return (
                "❌ SOLANA SCAN FAILED\n\n"
                f"{type(e).__name__}: {e}"
            )

    # -----------------------------------------------------
    # EVM ADDRESS
    # Check BNB and Robinhood automatically
    # -----------------------------------------------------

    if looks_like_evm(token):

        results = []

        for chain in ["bnb", "robinhood"]:

            print(
                f"🔎 Checking {chain.upper()}: {token}"
            )

            try:

                result = analyze_token(
                    chain,
                    token
                )

                results.append(
                    make_alert(result)
                )

            except Exception as e:

                print(
                    f"⚠️ {chain.upper()} not found:"
                    f" {type(e).__name__}: {e}"
                )

        if results:

            return "\n\n━━━━━━━━━━━━━━━━━━\n\n".join(
                results
            )

        return (
            "❌ TOKEN NOT FOUND\n\n"
            "This EVM contract did not return "
            "usable data on BNB Chain or Robinhood Chain."
        )

    return (
        "❌ INVALID CONTRACT ADDRESS\n\n"
        "Please paste only the token contract address."
    )


print("🤖 TEMMYDEEXC PASTE-ONLY SCANNER")
print("================================")
print("✅ Solana addresses: automatic")
print("✅ BNB addresses: automatic")
print("✅ Robinhood addresses: automatic")
print("🔐 Read-only")
print()
print("Paste a contract address into Telegram.")
```
